```python
#find h index
#input   citations[]=[3,0,5,3,0]  output=3
#in the output 1 is 5times repeted ...2 -3timesrepeated ...3-3times repeated
#4-1 time repeated and 5 -1 time
#1 minimum 1 should be repeated and 2 minimum 2 times...and soo on
#you are given an array of citations[],where each element citation[i]
#represents thenumber of citations received by the ith paper of a reseacher
#you have to caluclate the reseachers h-index
#the h-index is defined as the maxium vaue h,such that the reseacher has published  at least h papers
#and all those papers have citation value greater han h equal to h
citations=[3,0,5,3,0]
n=len(citations)
freq=[0]*(n-1)
for i in range(n):
    if citations[i]>=n:
        freq[n]+=1
    else:
        freq[citations[i]]+=1
s=freq[n]
while s<n:
    n-=1
    s+=freq[n]
print(n)
```

```python
citations = [3, 0, 5, 3, 0]
n = len(citations)

# frequency array of size n+1
freq = [0] * (n + 1)

# count frequencies
for c in citations:
    if c >= n:
        freq[n] += 1
    else:
        freq[c] += 1

# accumulate from the top down
s = 0
for i in range(n, -1, -1):
    s += freq[i]
    if s >= i:
        print(i)
        break
```

```python
class Solution(object):
    def twoSum(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """
        seen = {}
        for i, num in enumerate(nums):
            complement = target - num
            if complement in seen:
                return [seen[complement], i]
            seen[num] = i
```
